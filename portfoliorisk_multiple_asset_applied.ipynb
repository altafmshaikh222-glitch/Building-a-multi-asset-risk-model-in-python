```python
import pandas as pd
import numpy as np
import yfinance as yf
```

```python
# The 10 tickers shown in your image
tickers = ['AAPL', 'KO', 'NFLX', 'BRK-B', 'DIS', 'IBM', 'VZ', 'WMT', 'GE', 'TSLA']

# Download adjusted closing prices matching the timeframe from the image
# (Starts early Jan 2012)
downloaded = yf.download(
    tickers,
    start="2021-08-01",
    end="2026-09-01",
    auto_adjust=False,
    progress=False
)

# Use adjusted prices; fall back to Close if unavailable
if "Adj Close" in downloaded.columns.get_level_values(0):
    df = downloaded["Adj Close"]
else:
    df = downloaded["Close"]

# Format the index into the 'DD/MM/YYYY 16:00' timestamp format
df.index = df.index.strftime('%d/%m/%Y 16:00')

# Rename the index column name to match the screenshot
df.index.name = 'date_gsheets'

# Reorder columns to match the exact order in your image
df = df[tickers]

# Save to CSV
df.to_csv('10stocks_price.csv')

print("CSV created successfully!")
```

```python
returns_df = df.pct_change(1)
```

```python
num_of_stocks = 10
weights = [1/num_of_stocks] * num_of_stocks  # Equal weights for each stock
weights
```

```python
#Calculate the variance covariance matrix
vcv_matrix = returns_df.cov()
vcv_matrix
```

```python
var_p = np.dot(np.transpose(weights), np.dot(vcv_matrix, weights))
var_p
```

```python
sd_p = np.sqrt(var_p)
sd_p
```

```python
sd_p_annual = sd_p * np.sqrt(252)  # Assuming 252 trading days in a year
sd_p_annual
```

```python
# Calculate annualized individual standard deviation per column
individual_risks = returns_df.std() * np.sqrt(252)

# Display the series
individual_risks
```
